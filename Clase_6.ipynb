# Ciclo For y Ciclo while

## Ciclo for

```python
for i in range(10):
    
    print(i)
```

```python
alumnos = ["Juan", "Pedro", "Maria", "Luis", "Ana"]
for i in alumnos:
    print(i)
```

```python
for i in "Iteración":
    print(i)
```

```python
suma = 0
for i in range(1,101):
    suma += i
suma
```

```python
import random
for i in alumnos:
    print(f"Calificaciones de {i}: ")
    calificaciones = [random.randint(0,100) for j in range(3)]
    print(f"Matematicas: {calificaciones[0]}")
    print(f"Fisica: {calificaciones[1]}")
    print(f"Quimica: {calificaciones[2]}")
```

```python
#Break

for i in range(1,10):
    if i%2 == 0:
        print(f"{i} es par")
        break
```

## While

```python
bateria = 100
while bateria > 0:
    print("Tarea en ejecución...")
    print("Tarea realizada.")
    print(f"Bateria: {bateria}%")
    bateria -= 1 
```

```python
flag = True

while flag:
    print("Estas en un ciclo infinito ... olvidate romper el ciclo") 
    break
```

```python
flag = True

while flag==True:
    print("Estas en un ciclo infinito ... olvidate romper el ciclo") 
    flag = False
```

```python
flag = True

contador = 0

print(contador,flag)

while flag==True:
    print("Estas en un ciclo infinito ... olvidate romper el ciclo") 
    if contador == 100:
       flag = False
    contador += 1
print(contador,flag)
```
