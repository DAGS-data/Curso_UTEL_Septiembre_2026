```python
# Función sencilla.
def formula_general(a: float, b: float, c: float)-> list[float, float]:
    root_1 = (-b+(b**2-4*a*c)**(1/2))/(2*a)
    root_2 = (-b-(b**2-4*a*c)**(1/2))/(2*a)
    return root_1, root_2
```

```python
parabola = []
for x in range(-100, 100):
    parabola.append(5*x**2-2*x+1)
```

# Funciones con argumentos tupla (Args) y diccionario (Kwargs)

```python
def ingresa_datos(nombre: str, edad: int)-> list[str, int]:
    return nombre, edad     
```

```python
def ingresa_datos(*datos):
    lista = list(datos)
    return lista
```

```python
ingresa_datos("Juan", 25)
```

```python
ingresa_datos("Juan", 25, "CDMX","Programador","Perro","15,000 USD")
```

```python
def ingresa_datos(nombre: str, edad: int, *datos):
    lista = [nombre, edad] + list(datos)
    return lista
```

```python
ingresa_datos("Juan", 25, "CDMX","Programador","Perro","15,000 USD")
```

```python
def ingresa_datos(*datos):
    print(type(datos))
```

```python
ingresa_datos("Juan", 25, "Maria", 30, "Pedro", 22)
```

# Argumentos tipo **kwargs

```python
# Funcion con argumentos tipo **kwargs
def ingresa_datos(**kwargs):
    print((kwargs))
    return kwargs
```

```python
ingresa_datos(nombre="Juan", edad= 25,ciudad= "CDMX",ocupacion = "Programador",mascota = "Perro",salario = "15,000 USD")
```

```python
def combinacion_de_argumentos(correo: str, nombre: str, contraseña: str, *aleatorios, **datos_adicionales):
    lista = [correo, nombre, contraseña] + list(aleatorios) + list(datos_adicionales.items())
    aleatorios = sum(aleatorios)
    resumen = f"""
        Datos del usuario:
        nombre: {nombre}
        correo: {correo}
        contraseña: {contraseña}
        Datos adicionales: {datos_adicionales}
        Suma de los números aleatorios: {aleatorios}
    """   

    return resumen
```

```python
print(combinacion_de_argumentos("user@example.com", "John Doe", "password123", 5, 10, 15,genero="Masculino", ocupacion="Ingeniero", ciudad="New York", pais="USA"))
```
