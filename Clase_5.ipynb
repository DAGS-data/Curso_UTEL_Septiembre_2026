# Elif

```python
edad = int(input("Ingrese su edad: "))
```

```python
if edad >= 18:
    print("Debes ir a la universidad.")
if (edad <18 ) and (edad >= 15):
    print("Debes ir a la preparatoria.")
if (edad < 15) and (edad >= 6):
    print("Debes ir a la secundaria.")
else:
    print("Debes ir a la primaria.")
```

```python
if edad >= 18:
    print("Debes ir a la universidad.")
elif (edad <18 ) and (edad >= 15):
    print("Debes ir a la preparatoria.")
elif (edad < 15) and (edad >= 6):
    print("Debes ir a la secundaria.")
else:
    print("Debes ir a la primaria.")
```
